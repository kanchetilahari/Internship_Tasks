Q1. Find the Shortest Word in a Sentence


```python
sentence = "Python makes coding easy"
words=sentence.split()
shortest=words[0]
for word in words:
    if len(word)<len(shortest):
        shortest=word
print(shortest)

# first we Store the sentence in one variable
# after we Use split() to separate the words
# Assume the first word is the shortest
# Compare the remaining words with the current shortest word
# If a word is shorter, update the shortest word
```

 Q6. Count Vowels in a String

```python
s="Education"
c=0
v="aeiou"
for ch in s.lower():
    if ch in v:
        c+=1
print(c)

# Store the string in a variable
# Create a variable c=0
# Loop through every character
# Check whether the character is a vowel
# increase the count by one
```

Q7. Count Consonants in a String

```python
s="Python 123!"
c=0
v="aeiou"
for ch in s.lower():
    if ch.isalpha() and ch not in v:
        c+=1
print(c)


# Create a variable c=0
# Loop through each character
# Check whether the character is an alphabet use alpha()
# then check weather it is not a vowel
# increase the count if both conditions are true
```

Q8. Find the Most Frequent Character


```python
s="banana"
freq={}
for ch in s:
    if ch!=" ":
        if ch in freq:
            freq[ch]+=1
        else:
            freq[ch]=1
most=None
max_count=0
for char,count in freq.items():
    if count>max_count:
        max_count=count
        most=char
print(most)

# Create an empty dictionary to store character frequencies
# Loop through each character
# Count each character occurrences
# Find the character with the highest frequency
```

Q9. Find the Least Frequent Character


```python
text = "success"
freq = {}
for char in text:
    if char != " ":
        freq[char] = freq.get(char, 0) + 1
least = None
min_count = float("inf")
for char, count in freq.items():
    if count < min_count:
        min_count = count
        least = char
print(least)

# Create an empty dictionary to count each char
# store each character frequency
# Assume the first character is the least frequent
```

Q10. Find All Characters Appearing More Than Once


```python
text="programming"
freq={}
for char in text:
    if char!=" ":
        freq[char]=freq.get(char,0)+1
repeat=[]
for char in text:
    if freq[char]>1 and char not in repeat:
        repeat.append(char)
print(repeat)

# Create an empty dictionary to count characters and Count each character
# Create an empty list for repeated characters
# Loop through the original string
# If a character appears more than once and then append it
```

Q15. Create a Dictionary of Numbers and Their Even/Odd Status

```python
numbers=[1,2,3,4,5]
res={}
for num in numbers:
    if num%2==0:
        res[num]="Even" 
    else:
        res[num]="Odd"
print(res)

# Create an empty dictionary
# If num % 2 == 0, it is "Even" Otherwise "Odd"
```
