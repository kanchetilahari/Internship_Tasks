Q35

```python
import copy
a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)

# copy.copy(a) creates a new outer list here
# now Changing an element inside an inner list affects both a and b
```

Q36

```python
import copy
a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)

# copy.deepcopy(a) creates an independent copy of the outer list and its nested lists
#  here modifying b does not change a
```

Q37

```python
import copy
a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)

# here a.copy() creates a shallow copy
# he outer lists are different, but their inner lists are shared
# Appending 30 to a[1] also affects the corresponding inner list referenced by b
```

Q38

```python
import copy
a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)

# deepcopy() duplicates the nested lists
# and also adding 30 to the inner list in a does not affect b
```
